# Inicializacion

Se deja un bloque de código para la inicialización de las funciones e imports de bibliotecas necesarias.

```python
# Librerias de pyhon a utilizar
import numpy as np
import math
import IPython
from IPython.display import display, Math
import pandas as pd
```

```python
# Funciones Auxiliares
from typing import Any

def incert_A(mediciones):
  s = np.std(mediciones, ddof=1, )  # ddof=1 -> grados de libertad v = 1/n-1
  return s/math.sqrt(len(mediciones))

def incert_B(err_rel, nd, ctas, mediciones):
  return (err_rel/100 + nd/ctas) * np.mean(mediciones) / math.sqrt(3)

def incert_Total(uA, uB):
  return math.sqrt(uA**2 + uB**2)

def incert_expandida(uC, k):
  return k * uC
```

## Caso 2: Medición indirecta

Para este caso se medirá la tensión de salida partiendo de los valores de componentes y de la fuente medidos con el mismo multímetro.

Los datos obtenidos con el multímetro son:
$$R_A = 9.84 k\Omega$$
$$R_B = 9.86 k\Omega$$

$$V_F = 10.07 V$$

Para el caso de las resistencias se usó la escala de $60K\Omega$ y para la tensión la escala de $60V$.

Se obtendrá la incertidumbre de tipo B de cada medición, en este caso usando la expresión con la resolución para más simpleza:

$$\Delta X= \frac{\epsilon_r \%}{100} . X  + n.Res(X)$$

Siendo la resolución el cociente entre la escala y la cantidad de cuentas máximas de la escala. 

$$Res(X) = \frac{X}{c_X}$$

Luego, se obtiene la incertidumbre dividiendo en este caso por raiz de tres por ser una distribución rectangular.

$$\mu_B (X) = \frac{\Delta X}{\sqrt{3}}$$

Lo que se hará ahora es utilizando la ecuación del divisor resistivo obtener la tensión resultante. La ecuación es:

$$V_o = V_F . \frac{R_B}{R_A + R_B}$$

También hay que propagar todos los errores derivando para cada caso.



```python
escR = 60e3
cR = 6000
escV = 60
cV = 6000

#Valores medidos
RA = 9.84e3      # ohm
RB = 9.86e3      # ohm
VF = 10.07       # V

ResR = escR / cR 
ResV = escV / cV 

ci_RA = RA / ResR
ci_RB = RB / ResR
ci_VF = VF / ResV

print(ci_RA, ci_RB, ci_VF)

uB_RA_2 = incert_B(err_r, n_r, ci_RA, [RA])
uB_RB_2 = incert_B(err_r, n_r, ci_RB, [RB])
uB_VF_2 = incert_B(err_dc, n_dc, ci_VF, [VF])

print("uB(RA): ",uB_RA_2)
print("uB(RB): ",uB_RB_2)
print("uB(VF): ",uB_VF_2)
```

Habiendo obtenido las incertidumbres de los tres datos ahora debemos propagar la ecuación, para eso sigo la siguiente fórmula:

$$\Large \mu_c(V_o) = \sqrt{\sum_{i=1}^{N} \Biggl(
\frac
{\partial{V_o}}
{\partial{x_i}}  \, \mu(x_i) 
\Biggr)^2}$$

Aplicándolo sobre:
$$V_o = V_F . \frac{R_B}{R_A + R_B}$$

Se deben realizar las derivadas parciales de:

$$\frac{\partial{V_o}}{\partial{V_F}} = \frac{R_B}{R_A+R_B}$$
$$\frac{\partial{V_o}}{\partial{R_A}} = -\frac{V_F . R_B}{(R_A+R_B)^2}$$
$$\frac{\partial{V_o}}{\partial{R_B}} = \frac{V_F . R_A}{(R_A+R_B)^2}$$

Reemplazando en la fórmula antes mostrada:
$$\mu_c(V_o) = \sqrt{ 
\Biggl(\frac{\partial{V_o}}{\partial{V_F}}  . \mu(V_F) \Biggr)^2 + 
\Biggl(\frac{\partial{V_o}}{\partial{R_A}}  . \mu(R_A) \Biggr)^2 +
\Biggl(\frac{\partial{V_o}}{\partial{R_B}}  . \mu(R_B) \Biggr)^2 }
$$

$$\mu_c(V_o) = \sqrt{ 
\Biggl(\frac{R_B}{R_A+R_B}  . \mu(V_F) \Biggr)^2 + 
\Biggl(-\frac{V_F . R_B}{(R_A+R_B)^2}  . \mu(R_A) \Biggr)^2 +
\Biggl(\frac{V_F . R_A}{(R_A+R_B)^2}  . \mu(R_B) \Biggr)^2 }
$$

Se hace el procedimiento en python para obtener los resultados numéricos:

```python
# Derivadas parciales

dVo_dVF = RB/(RA+RB)
dVo_dRA = -(VF*RB)/(RA+RB)**2
dVo_dRB = (VF*RA)/(RA+RB)**2

# Incertidumbre Combinada
uC_Vo = np.sqrt(
    (dVo_dVF*uB_VF_2)**2 +
    (dVo_dRA*uB_RA_2)**2 +
    (dVo_dRB*uB_RB_2)**2
)

# Incertidumbre Expandida
U_Vo = k * uC_Vo
print("U(Vo) = ",U_Vo)
```

Como se ve en la simulación, se obtiene la incertidumbre de la medición:
$$U(Vo) = 0.06082066499967485 V $$

Y el resultado de hacer:
$$V_o = V_F . \frac{R_B}{R_A + R_B}$$
Con los valores medidos al principio del ejercicio $V_o$ vale:
$$V_o = 5.040111675 V$$

## Caso 3: Especificaciones

Para este estudio se medirá la tensión utilizando las especificaciones técnicas de los componentes.

En el caso de las resistencias los valores de las mismas se pueden hayar mediante el código de colores. Para este laboratorio utilizamos resistencias de $10K\Omega$, la tolerancia se obtuvo mediante la cuarta banda, que en este caso es dorada por lo que es al $5\%$.

Esto significa que, el $\Delta R$ se obtiene sin el segundo término de la ecuación ya que no hay un display que muestre los resultados, quedando:
$$\Delta R = (\frac{\epsilon_r \%}{100} + 0) . R$$

Obteniendo la incertidumbre final, teniendo una distribución rectangular:
$$\mu_B(R) = \frac{\Delta R}{\sqrt{3}}$$

Para el caso de la fuente, el modelo de la misma es HY3002D. Según la datasheet el error presentado en el display es de:
$$\pm (1\% + 2 \ dígitos)$$

En nuestro caso, en el display aparecía:
$$V_F = 10 V$$

Resolviendo:
$$\Delta V_F = (\frac{1 \%}{100} + \frac{2}{1000}) . V_F$$

$$\mu_B (V_F) = \frac{\Delta V_F}{\sqrt{3}}$$

$$U(V_F) = K_{95\%} . \mu_B(V_F)$$

Se realizan los cálculos en python:

```python
escV = 60
cV = 6000

RA_esp = 10e3
RB_esp = 10e3
VF = 10.00

ResV = escV / cV 
ci_VF = VF / ResV

uB_RA_3 = incert_B(5, 0, 1, [RA_esp])
uB_RB_3 = incert_B(5, 0, 1, [RB_esp])
uB_VF_3 = incert_B(err_fuente, n_fuente, ci_VF, [VF])

print("uB(RA): ",uB_RA_3, "Ω")
print("uB(RB): ",uB_RB_3, "Ω")
print("uB(VF): ",uB_VF_3, "V")
```

Con los resultados de las incertidumbres obtenidas, ahora hay que propagar. Para eso se reutiliza la propagación realizada para el Caso 2, ya que es la misma fórmula:
$$V_o = V_F . \frac{R_B}{R_A+R_B}$$

Luego de realizar las derivadas parciales, la propagación queda:
$$\mu_c(V_o) = \sqrt{ 
\Biggl(\frac{R_B}{R_A+R_B}  . \mu(V_F) \Biggr)^2 + 
\Biggl(-\frac{V_F . R_B}{(R_A+R_B)^2}  . \mu(R_A) \Biggr)^2 +
\Biggl(\frac{V_F . R_A}{(R_A+R_B)^2}  . \mu(R_B) \Biggr)^2 }
$$

```python
# Derivadas parciales

dVo_dVF = RB_esp/(RA_esp+RB_esp)
dVo_dRA = -(VF*RB_esp)/(RA_esp+RB_esp)**2
dVo_dRB = (VF*RA_esp)/(RA_esp+RB_esp)**2

# Incertidumbre Combinada
uC_Vo_3 = np.sqrt(
    (dVo_dVF*uB_VF_3)**2 +
    (dVo_dRA*uB_RA_3)**2 +
    (dVo_dRB*uB_RB_3)**2
)

# Incertidumbre Expandida
U_Vo_3 = k * uC_Vo_3
print("U(Vo) = ",U_Vo_3)

# Resultado del divisor resistivo
Vo_3 = VF * (RB_esp / (RA_esp+RB_esp))

print("Vo = ", Vo_3)
```

Finalmente, con los resultados obtenidos:
$$U(V_o) = 0.1778380583564722 V$$
$$V_o = 5.0 V$$
