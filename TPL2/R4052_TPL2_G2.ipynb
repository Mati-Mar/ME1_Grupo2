# Inicialización

Se deja un bloque de código para la inicialización de las funciones e imports de bibliotecas necesarias.

```python
# Librerías de Python a utilizar
import numpy as np
import math
import IPython
from IPython.display import display, Math
import pandas as pd
```

```python
# Funciones Auxiliares para cálculo de Incertidumbre
from typing import Any

def incert_A(mediciones):
  s = np.std(mediciones, ddof=1)  # ddof=1 -> grados de libertad v = n-1
  return s / math.sqrt(len(mediciones))

def incert_B(err_rel, nd, ctas, mediciones):
  return (err_rel / 100 + nd / ctas) * np.mean(mediciones) / math.sqrt(3)

def incert_Total(uA, uB):
  return math.sqrt(uA**2 + uB**2)

def incert_expandida(uC, k):
  return k * uC
```

# Procedimiento específico de la medición

## Descripción

A partir de la utilización de un divisor de tensión resistivo, se busca obtener la tensión de salida del mismo causando un error de método al empeorar artificialmente la resistencia voltimétrica ($R_V$) del instrumento de sensor.

El circuito de medición cuenta con una fuente de alimentación ($V_F$), su resistencia interna ($R_F$), los resistores del divisor de tensión ($R_A$ y $R_B$), y la resistencia interna del instrumento ($R_V$).

Se analizan dos circuitos o condiciones de medición para obtener dos errores de método diferentes:

### Caso 1: Error de método despreciable
En este caso la resistencia del voltímetro $R_V$ es lo suficientemente alta frente a la impedancia del divisor como para que la corriente derivada hacia el instrumento no afecte significativamente la medición.

### Caso 2: Error de método del 20%
En este caso se altera artificialmente la resistencia voltimétrica agregando una resistencia en paralelo de valor calculado para forzar una caída del 20% en la tensión medida respecto al valor teórico ideal.

# Resolución matemática y mediciones


## Caso 1: Error de método despreciable

Aquí deben realizar los cálculos teóricos y colocar el valor medido en el display.

La ecuación general del divisor de tensión considerando el instrumento en paralelo con $R_B$ es:

$$V_{medido} = V_F \cdot \frac{R_B \parallel R_V}{R_A + (R_B \parallel R_V)}$$

Como el error es despreciable, se asume que $R_V \gg R_B$.

```python
V_directa_1 = 5.000  # V
ci_1 = 5000         # Cuentas en el display

uC_V_1 = incert_B(err_dc, n_dc, ci_1, [V_directa_1])
U_V_1 = incert_expandida(uC_V_1, k)
print(f"Incertidumbre expandida U(V) Caso 1: {U_V_1:.5f} V")
```

## Caso 2: Error de método del 20%

Para forzar un error de método del 20%, la tensión medida debe ser el 80% de la tensión ideal. Para ello, se calcula el valor necesario del paralelo equivalente entre $R_B$ y la nueva $R_V$ modificada.

Coloquen aquí el desarrollo de las derivadas parciales correspondientes para la propagación de las incertidumbres en este escenario modificado.

```python
V_directa_2 = 4.050  # V
ci_2 = 4050         # Cuentas en el display

uC_V_2 = incert_B(err_dc, n_dc, ci_2, [V_directa_2])
U_V_2 = incert_expandida(uC_V_2, k)
print(f"Incertidumbre expandida U(V) Caso 2: {U_V_2:.5f} V")
```
