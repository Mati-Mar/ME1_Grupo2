## Procedimiento

### Medición de tensión

1. Se conectaron ambos voltímetros en paralelo con la fuente.
2. Se seleccionó en cada instrumento el rango correspondiente.
3. Se ajustó la fuente al valor nominal.
4. Se registraron simultáneamente las indicaciones del patrón y del instrumento a calibrar.
5. Se repitieron las mediciones para todos los puntos y rangos disponibles.

### Medición de corriente

1. Se conectaron ambos amperímetros en serie con la fuente y la carga.
2. Se verificaron los bornes y rangos antes de energizar.
3. Se ajustó la corriente al valor nominal.
4. Se registraron simultáneamente las indicaciones del patrón y del instrumento a calibrar.
5. Se repitió el procedimiento para los rangos disponibles.



## Metodología de cálculo utilizada por la cátedra

Para un conjunto de \(n\) mediciones \(X_i\), se calcula primero el valor medio:

$$
\overline{X}=\frac{1}{n}\sum_{i=1}^{n}X_i
$$

La desviación estándar experimental es:

$$
s(X)=\sqrt{\frac{1}{n-1}\sum_{i=1}^{n}\left(X_i-\overline{X}\right)^2}
$$

La incertidumbre tipo A es:

$$
u_A(X)=\frac{s(X)}{\sqrt{n}}
$$

Para un multímetro cuya exactitud se expresa como:

$$
\pm\left(er\%+N\ \text{dígitos}\right)
$$

el límite de error se calcula mediante:

$$
\Delta X=
\left(
\frac{er\%}{100}
+
\frac{N}{c_i}
\right)X_{\mathrm{IND}}
$$

De forma equivalente, utilizando la resolución \(r\):

$$
\Delta X=
\frac{er\%}{100}X_{\mathrm{IND}}+N\,r
$$

La incertidumbre tipo B es:

$$
u_B(X)=\frac{\Delta X}{\sqrt{3}}
$$

La incertidumbre combinada es:

$$
u_C(X)=\sqrt{u_A^2(X)+u_B^2(X)}
$$

La incertidumbre expandida, con \(K=2\), es:

$$
U(X)=K\,u_C(X)
$$

Finalmente, el resultado se presenta como:

$$
X=\left(\overline{X}\pm U(X)\right)
\qquad K=2
$$

## Datos de exactitud y resolución

Las funciones siguientes implementan las especificaciones de los dos instrumentos.

Para el **UT890D** se utilizan exactamente las especificaciones suministradas para cada rango.

Para el **GDM-8145** se emplean las especificaciones del equipo patrón cargadas en el trabajo:

- Tensión continua: ±(0,03 % de la lectura + 4 dígitos).
- Corriente continua hasta 200 mA: ±(0,2 % de la lectura + 2 dígitos).
- Rangos de 2 A y 20 A: ±(0,3 % de la lectura + 2 dígitos).


```python
# Resoluciones del patrón GDM-8145, expresadas en V o A.
RES_GDM_V = {
    "200mV": 10e-6,
    "2V": 100e-6,
    "20V": 1e-3,
    "200V": 10e-3,
    "1000V": 100e-3,
    "1200V": 100e-3,   # Nombre utilizado en la planilla.
}

RES_GDM_I = {
    "200uA": 10e-9,
    "200µA": 10e-9,
    "2mA": 0.1e-6,
    "20mA": 1e-6,
    "200mA": 10e-6,
    "2000mA": 100e-6,
    "2A": 100e-6,
    "20A": 1e-3,
}

# Resoluciones y exactitudes del UT890D.
UT_V = {
    "600mV": {"res": 0.1e-3, "er": 0.5, "dig": 4},
    "6V":     {"res": 1e-3,   "er": 0.5, "dig": 2},
    "60V":    {"res": 10e-3,  "er": 0.5, "dig": 2},
    "600V":   {"res": 0.1,    "er": 0.5, "dig": 2},
    "1000V":  {"res": 1.0,    "er": 0.7, "dig": 10},
}

UT_I = {
    "60uA":   {"res": 0.01e-6, "er": 0.8, "dig": 8},
    "60µA":   {"res": 0.01e-6, "er": 0.8, "dig": 8},
    "6mA":    {"res": 0.001e-3, "er": 0.8, "dig": 8},
    "60mA":   {"res": 0.01e-3,  "er": 0.8, "dig": 8},
    "600mA":  {"res": 0.1e-3,   "er": 1.2, "dig": 5},
    "20A":    {"res": 0.01,      "er": 2.0, "dig": 5},
}

def media(valores):
    return float(np.mean(np.asarray(valores, dtype=float)))

def desviacion_estandar(valores):
    valores = np.asarray(valores, dtype=float)
    if len(valores) <= 1:
        return 0.0
    return float(np.std(valores, ddof=1))

def incertidumbre_tipo_a(valores):
    valores = np.asarray(valores, dtype=float)
    if len(valores) <= 1:
        return 0.0
    return desviacion_estandar(valores) / math.sqrt(len(valores))

def parametros_gdm(magnitud, escala):
    if magnitud == "V":
        return {"res": RES_GDM_V[escala], "er": 0.03, "dig": 4}

    res = RES_GDM_I[escala]
    if escala in ("2000mA", "2A", "20A"):
        return {"res": res, "er": 0.3, "dig": 2}
    return {"res": res, "er": 0.2, "dig": 2}

def parametros_ut(magnitud, escala):
    tabla = UT_V if magnitud == "V" else UT_I
    return tabla[escala]

def limite_error(valor_indicado, er_porcentual, numero_digitos, resolucion):
    return (
        (er_porcentual / 100.0) * abs(valor_indicado)
        + numero_digitos * resolucion
    )

def incertidumbre_tipo_b(delta):
    return delta / math.sqrt(3)

def incertidumbre_combinada(u_a, u_b):
    return math.sqrt(u_a**2 + u_b**2)

def incertidumbre_expandida(u_c, k=K):
    return k * u_c
```

## Cálculo para cada punto y rango

Se aplica por separado la metodología completa al patrón y al UT890D:

1. $$\overline{X}$$
2. $$s(X)$$
3. $$u_A(X)$$
4. $$\Delta X$$
5. $$u_B(X)$$
6. $$u_C(X)$$
7. $$U(X)$$

La desviación de indicación del instrumento a calibrar respecto del patrón se define como:

$$
D=\overline{X}_{UT}-\overline{X}_{P}
$$

Para completar la columna **Verifica**, se compara el intervalo del resultado del patrón con el intervalo declarado para el instrumento a calibrar:

$$
|D|+U(P)\leq U(UT)
$$

Este criterio expresa que la desviación observada, considerando la incertidumbre expandida del patrón, debe quedar contenida dentro de la incertidumbre expandida asignada al UT890D.


```python
resultados = []
calculos = []

for g in grupos:
    magnitud = g["magnitud"]
    xp = np.asarray(g["patron"], dtype=float)
    xi = np.asarray(g["instrumento"], dtype=float)

    # Patrón
    media_p = media(xp)
    s_p = desviacion_estandar(xp)
    uA_p = incertidumbre_tipo_a(xp)
    par_p = parametros_gdm(magnitud, g["escala_patron"])
    delta_p = limite_error(media_p, par_p["er"], par_p["dig"], par_p["res"])
    uB_p = incertidumbre_tipo_b(delta_p)
    uC_p = incertidumbre_combinada(uA_p, uB_p)
    U_p = incertidumbre_expandida(uC_p)

    # Instrumento a calibrar
    media_i = media(xi)
    s_i = desviacion_estandar(xi)
    uA_i = incertidumbre_tipo_a(xi)
    par_i = parametros_ut(magnitud, g["escala_instrumento"])
    delta_i = limite_error(media_i, par_i["er"], par_i["dig"], par_i["res"])
    uB_i = incertidumbre_tipo_b(delta_i)
    uC_i = incertidumbre_combinada(uA_i, uB_i)
    U_i = incertidumbre_expandida(uC_i)

    D = media_i - media_p
    verifica = "Sí" if abs(D) + U_p <= U_i else "No"

    calculos.append({
        "Magnitud": "Tensión" if magnitud == "V" else "Corriente",
        "Valor nominal": g["valor_nominal"],
        "Escala patrón": g["escala_patron"],
        "Escala UT890D": g["escala_instrumento"],
        "n": len(xp),
        "Media patrón": media_p,
        "s(P)": s_p,
        "uA(P)": uA_p,
        "Delta(P)": delta_p,
        "uB(P)": uB_p,
        "uC(P)": uC_p,
        "U(P)": U_p,
        "Media UT890D": media_i,
        "s(UT)": s_i,
        "uA(UT)": uA_i,
        "Delta(UT)": delta_i,
        "uB(UT)": uB_i,
        "uC(UT)": uC_i,
        "U(UT)": U_i,
        "D": D,
        "Verifica": verifica,
    })

    # Resultado final pedido por el TP.
    resultados.append({
        "Magnitud": "Tensión" if magnitud == "V" else "Corriente",
        "Valor nominal": g["valor_nominal"],
        "Escala": g["escala_instrumento"],
        "Valor verdadero": media_p,
        "Valor medido": media_i,
        "Punto calibrado": media_i,
        "Factor de expansión": K,
        "Desviación": D,
        "Verifica": verifica,
    })

df_calculos = pd.DataFrame(calculos)
df_resultados = pd.DataFrame(resultados)

display(df_calculos)
```

## Presentación de resultados

Para evitar ambigüedades:

- **Valor verdadero:** promedio indicado por el patrón GDM-8145.
- **Valor medido:** promedio indicado por el UT890D.
- **Punto calibrado:** indicación promedio del UT890D correspondiente al punto y rango ensayado.
- **Desviación:** $$D=\overline{X}_{UT}-\overline{X}_{P}$$.
- **Factor de expansión:** \(K=2\).


```python
def formatear_tabla(df, magnitud):
    salida = df.copy()

    if magnitud == "Tensión":
        def fmt(x):
            if abs(x) < 1:
                return f"{x*1000:.3f} mV"
            return f"{x:.5f} V"
    else:
        def fmt(x):
            return f"{x*1000:.5f} mA"

    for columna in ["Valor verdadero", "Valor medido", "Punto calibrado", "Desviación"]:
        salida[columna] = salida[columna].map(fmt)

    salida["Factor de expansión"] = salida["Factor de expansión"].map(
        lambda x: f"{x:.0f}"
    )
    return salida

columnas_tp = [
    "Valor nominal",
    "Escala",
    "Valor verdadero",
    "Valor medido",
    "Punto calibrado",
    "Factor de expansión",
    "Desviación",
    "Verifica",
]

tabla_tension = formatear_tabla(
    df_resultados[df_resultados["Magnitud"] == "Tensión"][columnas_tp],
    "Tensión",
).reset_index(drop=True)

tabla_corriente = formatear_tabla(
    df_resultados[df_resultados["Magnitud"] == "Corriente"][columnas_tp],
    "Corriente",
).reset_index(drop=True)

print("TENSIÓN CONTINUA")
display(tabla_tension.style.hide(axis="index"))

print("CORRIENTE CONTINUA")
display(tabla_corriente.style.hide(axis="index"))
```

## Resultados completos con incertidumbre

A continuación se muestran los resultados de cada instrumento con la notación utilizada por la cátedra:

$$
X=\left(\overline{X}\pm U(X)\right)
\qquad K=2
$$


```python
def resultado_con_incertidumbre(media_x, U_x, magnitud):
    if magnitud == "Tensión":
        if max(abs(media_x), abs(U_x)) < 1:
            return f"({media_x*1000:.3f} ± {U_x*1000:.3f}) mV, K = 2"
        return f"({media_x:.5f} ± {U_x:.5f}) V, K = 2"

    return f"({media_x*1000:.5f} ± {U_x*1000:.5f}) mA, K = 2"

presentacion = []
for _, fila in df_calculos.iterrows():
    presentacion.append({
        "Magnitud": fila["Magnitud"],
        "Valor nominal": fila["Valor nominal"],
        "Escala patrón": fila["Escala patrón"],
        "Resultado patrón": resultado_con_incertidumbre(
            fila["Media patrón"], fila["U(P)"], fila["Magnitud"]
        ),
        "Escala UT890D": fila["Escala UT890D"],
        "Resultado UT890D": resultado_con_incertidumbre(
            fila["Media UT890D"], fila["U(UT)"], fila["Magnitud"]
        ),
        "Verifica": fila["Verifica"],
    })

df_presentacion = pd.DataFrame(presentacion)
display(df_presentacion.style.hide(axis="index"))
```
