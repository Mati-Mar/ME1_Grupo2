## Metodología de cálculo utilizada por la cátedra

Para un conjunto de \(n\) mediciones \(X_i\), se calcula primero el valor medio:

$$
\overline{X}=\frac{1}{n}\sum_{i=1}^{n}X_i
$$

La desviación estándar experimental es:

$$
s(X)=\sqrt{\frac{1}{n-1}\sum_{i=1}^{n}\left(X_i-\overline{X}\right)^2}
$$

La incertidumbre tipo A es:

$$
u_A(X)=\frac{s(X)}{\sqrt{n}}
$$

Para un multímetro cuya exactitud se expresa como:

$$
\pm\left(er\%+N\ \text{dígitos}\right)
$$

el límite de error se calcula mediante:

$$
\Delta X=
\left(
\frac{er\%}{100}
+
\frac{N}{c_i}
\right)X_{\mathrm{IND}}
$$

De forma equivalente, utilizando la resolución \(r\):

$$
\Delta X=
\frac{er\%}{100}X_{\mathrm{IND}}+N\,r
$$

La incertidumbre tipo B es:

$$
u_B(X)=\frac{\Delta X}{\sqrt{3}}
$$

La incertidumbre combinada es:

$$
u_C(X)=\sqrt{u_A^2(X)+u_B^2(X)}
$$

La incertidumbre expandida, con \(K=2\), es:

$$
U(X)=K\,u_C(X)
$$

Finalmente, el resultado se presenta como:

$$
X=\left(\overline{X}\pm U(X)\right)
\qquad K=2
$$


```python
import math
import numpy as np
import pandas as pd
from IPython.display import display

K = 2.0

UNIDADES = {"V": 1.0, "mV": 1e-3, "mA": 1e-3}

# Mediciones del TP3: (magnitud, valor nominal, escala patrón, escala UT890D, unidad, lecturas patrón, lecturas UT890D)
mediciones = [
    ("V", "100mV", "200mV", "600mV", "mV",
     [101.58, 101.33, 101.56, 101.44, 101.36, 101.44, 101.46, 101.52, 101.60, 101.44, 101.43],
     [101.7, 101.4, 101.7, 101.5, 101.4, 101.4, 101.5, 101.5, 101.6, 101.5, 101.5]),
    ("V", "100mV", "2V", "6V", "mV",
     [101.4, 101.5, 101.4, 101.4, 101.5, 101.5, 101.4, 101.4, 101.5, 101.4, 101.5],
     [101, 101, 101, 101, 101, 101, 101, 101, 101, 101, 101]),
    ("V", "1V", "2V", "6V", "V",
     [1.0010, 1.0004, 1.0001, 0.9991, 0.9998, 0.9993, 0.9991, 0.9989, 0.9989, 0.9987, 0.9986],
     [1.001, 1.001, 1.001, 1.000, 1.000, 0.999, 0.999, 0.999, 0.999, 0.999, 0.999]),
    ("V", "1V", "20V", "60V", "V",
     [0.994, 0.995, 0.994, 0.995, 0.994, 0.994, 0.995, 0.995, 0.994, 0.994, 0.993],
     [0.99, 0.99, 0.99, 0.99, 0.99, 0.99, 0.99, 0.99, 0.99, 0.99, 0.99]),
    ("V", "10V", "20V", "60V", "V",
     [10.011, 10.009, 10.007, 10.005, 10.004, 10.004, 10.005, 10.003, 10.002, 10.000, 9.999],
     [10.01, 10.01, 10.00, 10.01, 10.00, 10.00, 10.00, 10.00, 10.00, 10.00, 10.00]),
    ("V", "10V", "1200V", "1000V", "V",
     [10.0, 10.0, 10.0, 10.0, 10.0, 10.0, 10.0, 10.0, 10.0, 10.0, 10.0],
     [9, 9, 9, 9, 9, 9, 9, 9, 9, 9, 9]),
    ("I", "1mA", "2mA", "6mA", "mA",
     [0.9942, 0.9941, 0.9937, 0.9934, 0.9933, 0.9932, 0.9931, 0.9929, 0.9929, 0.9928],
     [0.995, 0.995, 0.994, 0.994, 0.994, 0.994, 0.994, 0.994, 0.994, 0.994]),
    ("I", "1mA", "20mA", "60mA", "mA",
     [1.001, 1.002, 1.000, 1.001, 1.001, 1.001, 1.001, 1.001, 1.001, 1.002],
     [0.99, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00]),
    ("I", "10mA", "20mA", "60mA", "mA",
     [10.057, 10.069, 10.085, 10.050, 10.035, 10.037, 10.033, 10.032, 10.031, 10.029],
     [10.05, 10.05, 10.07, 10.06, 10.04, 10.04, 10.03, 10.03, 10.03, 10.03]),
    ("I", "10mA", "200mA", "600mA", "mA",
     [10.27, 10.28, 10.28, 10.28, 10.28, 10.27, 10.27, 10.27, 10.28, 10.28],
     [10.2, 10.2, 10.2, 10.2, 10.2, 10.2, 10.2, 10.2, 10.2, 10.2]),
    ("I", "100mA", "200mA", "600mA", "mA",
     [95.36, 95.78, 95.36, 95.78, 95.36, 95.36, 95.78, 95.36, 95.78, 95.78],
     [96.4, 95.4, 96.4, 95.4, 96.4, 95.4, 96.4, 95.4, 96.4, 95.4]),
    ("I", "180mA", "2000mA", "20A", "mA",
     [183.0, 182.8, 182.8, 182.8, 182.9, 182.9, 182.9, 182.8, 182.9, 183.0],
     [170, 170, 170, 170, 170, 170, 170, 170, 170, 170]),
]

# Conversión a unidades base (V y A).
grupos = [
    {
        "magnitud": mag,
        "valor_nominal": nominal,
        "escala_patron": esc_p,
        "escala_instrumento": esc_i,
        "unidad": unidad,
        "patron": [x * UNIDADES[unidad] for x in lect_p],
        "instrumento": [x * UNIDADES[unidad] for x in lect_i],
    }
    for mag, nominal, esc_p, esc_i, unidad, lect_p, lect_i in mediciones
]
```

## Valores medidos

```python
def tabla_medidos(magnitud):
    columnas = {}
    for mag, nominal, esc_p, esc_i, unidad, lect_p, lect_i in mediciones:
        if mag != magnitud:
            continue
        columnas[(f"{nominal} [{unidad}]", f"Patrón ({esc_p})")] = pd.Series(lect_p)
        columnas[(f"{nominal} [{unidad}]", f"UT890D ({esc_i})")] = pd.Series(lect_i)
    df = pd.DataFrame(columnas)
    df.index = range(1, len(df) + 1)
    df.index.name = "Lectura"
    return df.astype(object).fillna("")

print("TENSIÓN CONTINUA")
display(tabla_medidos("V"))

print("CORRIENTE CONTINUA")
display(tabla_medidos("I"))
```

## Datos de exactitud y resolución

Las funciones siguientes implementan las especificaciones de los dos instrumentos.

Para el **UT890D** se utilizan exactamente las especificaciones suministradas para cada rango.

Para el **GDM-8145** se emplean las especificaciones del equipo patrón cargadas en el trabajo:

- Tensión continua: ±(0,03 % de la lectura + 4 dígitos).
- Corriente continua hasta 200 mA: ±(0,2 % de la lectura + 2 dígitos).
- Rangos de 2 A y 20 A: ±(0,3 % de la lectura + 2 dígitos).


```python
# Resoluciones del patrón GDM-8145, expresadas en V o A.
RES_GDM_V = {
    "200mV": 10e-6,
    "2V": 100e-6,
    "20V": 1e-3,
    "200V": 10e-3,
    "1000V": 100e-3,
    "1200V": 100e-3,   # Nombre utilizado en la planilla.
}

RES_GDM_I = {
    "200uA": 10e-9,
    "200µA": 10e-9,
    "2mA": 0.1e-6,
    "20mA": 1e-6,
    "200mA": 10e-6,
    "2000mA": 100e-6,
    "2A": 100e-6,
    "20A": 1e-3,
}

# Resoluciones y exactitudes del UT890D.
UT_V = {
    "600mV": {"res": 0.1e-3, "er": 0.5, "dig": 4},
    "6V":     {"res": 1e-3,   "er": 0.5, "dig": 2},
    "60V":    {"res": 10e-3,  "er": 0.5, "dig": 2},
    "600V":   {"res": 0.1,    "er": 0.5, "dig": 2},
    "1000V":  {"res": 1.0,    "er": 0.7, "dig": 10},
}

UT_I = {
    "60uA":   {"res": 0.01e-6, "er": 0.8, "dig": 8},
    "60µA":   {"res": 0.01e-6, "er": 0.8, "dig": 8},
    "6mA":    {"res": 0.001e-3, "er": 0.8, "dig": 8},
    "60mA":   {"res": 0.01e-3,  "er": 0.8, "dig": 8},
    "600mA":  {"res": 0.1e-3,   "er": 1.2, "dig": 5},
    "20A":    {"res": 0.01,      "er": 2.0, "dig": 5},
}

def media(valores):
    return float(np.mean(np.asarray(valores, dtype=float)))

def desviacion_estandar(valores):
    valores = np.asarray(valores, dtype=float)
    if len(valores) <= 1:
        return 0.0
    return float(np.std(valores, ddof=1))

def incertidumbre_tipo_a(valores):
    valores = np.asarray(valores, dtype=float)
    if len(valores) <= 1:
        return 0.0
    return desviacion_estandar(valores) / math.sqrt(len(valores))

def parametros_gdm(magnitud, escala):
    if magnitud == "V":
        return {"res": RES_GDM_V[escala], "er": 0.03, "dig": 4}

    res = RES_GDM_I[escala]
    if escala in ("2000mA", "2A", "20A"):
        return {"res": res, "er": 0.3, "dig": 2}
    return {"res": res, "er": 0.2, "dig": 2}

def parametros_ut(magnitud, escala):
    tabla = UT_V if magnitud == "V" else UT_I
    return tabla[escala]

def limite_error(valor_indicado, er_porcentual, numero_digitos, resolucion):
    return (
        (er_porcentual / 100.0) * abs(valor_indicado)
        + numero_digitos * resolucion
    )

def incertidumbre_tipo_b(delta):
    return delta / math.sqrt(3)

def incertidumbre_combinada(u_a, u_b):
    return math.sqrt(u_a**2 + u_b**2)

def incertidumbre_expandida(u_c, k=K):
    return k * u_c
```

## Cálculo del punto calibrado

Para el patrón y para el UT890D se calcula $\overline{X}$, $s(X)$, $u_A$, $\Delta X$, $u_B$, $u_C$ y $U$ como se indicó antes.

La calibración transfiere la incertidumbre del patrón al instrumento. La incertidumbre de calibración combina la del patrón con la dispersión y la resolución $r$ del UT890D:

$$
u_{res}=\frac{r}{2\sqrt{3}}
\qquad
u_{cal}=\sqrt{u_C^2(V_p)+u_A^2(V_m)+u_{res}^2}
\qquad
U_{cal}=K\,u_{cal},\;K=2
$$

$$
\text{Punto calibrado}=\left(V_m\pm U_{cal}\right)
\qquad
\Delta v=V_m-V_p
$$

**Verifica:** el punto verifica la hoja de datos si la desviación más la incertidumbre de calibración queda dentro del semirrango del fabricante $\Delta_{fab}=\frac{er\%}{100}V_m+N\,r$:

$$
|\Delta v|+U_{cal}\leq\Delta_{fab}
$$


```python
def redondear(x, U):
    """Redondea U a 2 cifras significativas y x al mismo decimal."""
    dec = max(0, 1 - math.floor(math.log10(abs(U))))
    return f"{x:.{dec}f}", f"{U:.{dec}f}", dec

resultados = []
calculos = []

for g in grupos:
    magnitud = g["magnitud"]

    # Punto pendiente de medición: se deja el lugar en la tabla final.
    if len(g["patron"]) == 0 or len(g["instrumento"]) == 0:
        resultados.append({
            "Magnitud": "Tensión" if magnitud == "V" else "Corriente",
            "Valor nominal": g["valor_nominal"],
            "Escala": g["escala_instrumento"],
            "Valor verdadero": "—",
            "Valor medido": "—",
            "Punto calibrado": "—",
            "Factor de expansión": f"{K:.0f}",
            "Desviación": "—",
            "Verifica": "Pendiente",
        })
        continue
    xp = np.asarray(g["patron"], dtype=float)
    xi = np.asarray(g["instrumento"], dtype=float)

    # Patrón
    media_p = media(xp)
    s_p = desviacion_estandar(xp)
    uA_p = incertidumbre_tipo_a(xp)
    par_p = parametros_gdm(magnitud, g["escala_patron"])
    delta_p = limite_error(media_p, par_p["er"], par_p["dig"], par_p["res"])
    uB_p = incertidumbre_tipo_b(delta_p)
    uC_p = incertidumbre_combinada(uA_p, uB_p)
    U_p = incertidumbre_expandida(uC_p)

    # Instrumento a calibrar
    media_i = media(xi)
    s_i = desviacion_estandar(xi)
    uA_i = incertidumbre_tipo_a(xi)
    par_i = parametros_ut(magnitud, g["escala_instrumento"])
    delta_i = limite_error(media_i, par_i["er"], par_i["dig"], par_i["res"])  # semirrango del fabricante
    uB_i = incertidumbre_tipo_b(delta_i)
    uC_i = incertidumbre_combinada(uA_i, uB_i)
    U_i = incertidumbre_expandida(uC_i)

    # Punto calibrado
    u_res_i = par_i["res"] / (2 * math.sqrt(3))
    u_cal = math.sqrt(uC_p**2 + uA_i**2 + u_res_i**2)
    U_cal = K * u_cal

    D = media_i - media_p
    verifica = "Sí" if abs(D) + U_cal <= delta_i else "No"

    calculos.append({
        "Magnitud": "Tensión" if magnitud == "V" else "Corriente",
        "Valor nominal": g["valor_nominal"],
        "Escala patrón": g["escala_patron"],
        "Escala UT890D": g["escala_instrumento"],
        "n": len(xp),
        "Media patrón": media_p,
        "s(P)": s_p,
        "uA(P)": uA_p,
        "Delta(P)": delta_p,
        "uB(P)": uB_p,
        "uC(P)": uC_p,
        "U(P)": U_p,
        "Media UT890D": media_i,
        "s(UT)": s_i,
        "uA(UT)": uA_i,
        "Delta(UT)": delta_i,
        "uB(UT)": uB_i,
        "uC(UT)": uC_i,
        "U(UT)": U_i,
        "u_res(UT)": u_res_i,
        "u_cal": u_cal,
        "U_cal": U_cal,
        "D": D,
        "|D|+U_cal": abs(D) + U_cal,
        "Verifica": verifica,
    })

    # Resultado final con el formato de la consigna, en la unidad de cada punto.
    f = UNIDADES[g["unidad"]]
    u = g["unidad"]
    vm_txt, Ucal_txt, dec = redondear(media_i / f, U_cal / f)
    resultados.append({
        "Magnitud": "Tensión" if magnitud == "V" else "Corriente",
        "Valor nominal": g["valor_nominal"],
        "Escala": g["escala_instrumento"],
        "Valor verdadero": f"{media_p / f:.{dec}f} {u}",
        "Valor medido": f"{vm_txt} {u}",
        "Punto calibrado": f"({vm_txt} ± {Ucal_txt}) {u}",
        "Factor de expansión": f"{K:.0f}",
        "Desviación": f"{D / f:.{dec}f} {u}",
        "Verifica": verifica,
    })

df_calculos = pd.DataFrame(calculos)
df_resultados = pd.DataFrame(resultados)

pd.set_option("display.max_columns", None)
display(df_calculos)
```

- **Valor medido ($V_m$, $A_m$):** media del UT890D en cada escala.
- **Punto calibrado:** $V_m\pm U_{cal}$, con $K=2$.
- **Desviación:** $\Delta v=V_m-V_p$.
- **Verifica:** Sí si $|\Delta v|+U_{cal}\leq\Delta_{fab}$.


```python
def tabla_final(magnitud):
    v = "V" if magnitud == "Tensión" else "A"
    d = "v" if magnitud == "Tensión" else "i"
    df = df_resultados[df_resultados["Magnitud"] == magnitud].drop(columns="Magnitud")
    df = df.rename(columns={
        "Valor verdadero": f"Valor verdadero ({v}p)",
        "Valor medido": f"Valor medido ({v}m)",
        "Desviación": f"Desviación (Δ{d} = {v}m − {v}p)",
    })
    return df.set_index(["Valor nominal", "Escala"])

print("TENSIÓN CONTINUA")
display(tabla_final("Tensión"))

print("CORRIENTE CONTINUA")
display(tabla_final("Corriente"))
```
